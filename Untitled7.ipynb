```python
import torch
import torch.nn as nn


class FTTransformer(nn.Module):
    def __init__(
        self,
        num_features=11,
        d_token=64,
        n_heads=4,
        n_layers=3,
        dropout=0.2
    ):
        super().__init__()

        self.num_features = num_features
        self.d_token = d_token

        # Each numerical feature gets its own learnable embedding
        self.feature_embeddings = nn.Parameter(
            torch.randn(num_features, d_token) * 0.02
        )

        # CLS token used for final classification
        self.cls_token = nn.Parameter(
            torch.randn(1, 1, d_token) * 0.02
        )

        # Transformer encoder
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_token,
            nhead=n_heads,
            dim_feedforward=d_token * 4,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=n_layers
        )

        # Classification head
        self.norm = nn.LayerNorm(d_token)

        self.classifier = nn.Sequential(
            nn.Linear(d_token, 32),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(32, 1)
        )

    def forward(self, x):

        # x shape: (batch, num_features)

        # Convert each feature into a token
        x = x.unsqueeze(-1) * self.feature_embeddings.unsqueeze(0)

        # x shape: (batch, num_features, d_token)

        batch_size = x.size(0)

        # Add CLS token
        cls = self.cls_token.expand(batch_size, -1, -1)

        x = torch.cat([cls, x], dim=1)

        # Transformer
        x = self.transformer(x)

        # Take CLS representation
        x = x[:, 0]

        x = self.norm(x)

        # Binary classification
        output = self.classifier(x)

        return output.squeeze(1)


print("FT-Transformer architecture created successfully!")
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = FTTransformer(
    num_features=11,
    d_token=64,
    n_heads=4,
    n_layers=3,
    dropout=0.2
).to(device)

print("Device:", device)
print(model)
```

```python
from sklearn.model_selection import train_test_split
from torch.utils.data import TensorDataset, DataLoader

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_tensor,
    y_train_tensor,
    test_size=0.20,
    random_state=42,
    stratify=y_train_tensor
)

train_dataset = TensorDataset(X_tr, y_tr)
val_dataset = TensorDataset(X_val, y_val)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=64,
    shuffle=False
)

print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
```

```python
import copy
import numpy as np
from sklearn.metrics import roc_auc_score

num_epochs = 100
patience = 12

best_val_loss = float("inf")
best_model_state = None
patience_counter = 0

train_losses = []
val_losses = []
val_aucs = []

for epoch in range(num_epochs):

    # -------------------------
    # Training
    # -------------------------
    model.train()

    running_train_loss = 0.0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        outputs = model(X_batch)

        loss = criterion(outputs, y_batch)

        loss.backward()

        # Prevent excessively large gradients
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        running_train_loss += loss.item() * X_batch.size(0)

    train_loss = running_train_loss / len(train_loader.dataset)

    # -------------------------
    # Validation
    # -------------------------
    model.eval()

    running_val_loss = 0.0
    val_probabilities = []
    val_targets = []

    with torch.no_grad():

        for X_batch, y_batch in val_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            outputs = model(X_batch)

            loss = criterion(outputs, y_batch)

            running_val_loss += loss.item() * X_batch.size(0)

            probabilities = torch.sigmoid(outputs)

            val_probabilities.extend(
                probabilities.cpu().numpy()
            )

            val_targets.extend(
                y_batch.cpu().numpy()
            )

    val_loss = running_val_loss / len(val_loader.dataset)

    val_auc = roc_auc_score(
        val_targets,
        val_probabilities
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    val_aucs.append(val_auc)

    scheduler.step(val_loss)

    # -------------------------
    # Early stopping
    # -------------------------
    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_model_state = copy.deepcopy(
            model.state_dict()
        )

        patience_counter = 0

    else:

        patience_counter += 1

    if (epoch + 1) % 5 == 0 or epoch == 0:

        print(
            f"Epoch [{epoch+1:03d}/{num_epochs}] "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val AUC: {val_auc:.4f}"
        )

    if patience_counter >= patience:

        print(
            f"\nEarly stopping at epoch {epoch+1}"
        )

        break


# Restore best model
model.load_state_dict(best_model_state)

print("\nBest validation loss:", best_val_loss)
```

```python
import copy
from torch.utils.data import DataLoader


def create_model():
    return FTTransformer(
        num_features=11,
        d_token=64,
        n_heads=4,
        n_layers=3,
        dropout=0.2
    ).to(device)


def train_local_model(
    global_model,
    client_dataset,
    local_epochs=3,
    batch_size=32,
    learning_rate=0.001
):
    # Create a local copy of the global model
    local_model = copy.deepcopy(global_model)

    local_loader = DataLoader(
        client_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    # Calculate client-specific class weight
    labels = client_dataset.tensors[1]

    num_positive = (labels == 1).sum().item()
    num_negative = (labels == 0).sum().item()

    client_pos_weight = torch.tensor(
        [num_negative / num_positive],
        dtype=torch.float32
    ).to(device)

    local_criterion = nn.BCEWithLogitsLoss(
        pos_weight=client_pos_weight
    )

    local_optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    local_model.train()

    for epoch in range(local_epochs):

        for X_batch, y_batch in local_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            local_optimizer.zero_grad()

            outputs = local_model(X_batch)

            loss = local_criterion(
                outputs,
                y_batch
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                local_model.parameters(),
                max_norm=1.0
            )

            local_optimizer.step()

    return local_model
```

```python
def fedavg(global_model, client_models, client_sizes):

    global_state = global_model.state_dict()

    total_samples = sum(client_sizes)

    # Start with zeros
    averaged_state = {
        key: torch.zeros_like(value)
        for key, value in global_state.items()
    }

    # Weighted average
    for client_model, client_size in zip(
        client_models,
        client_sizes
    ):

        client_state = client_model.state_dict()

        weight = client_size / total_samples

        for key in averaged_state:

            averaged_state[key] += (
                client_state[key] * weight
            )

    # Update global model
    global_model.load_state_dict(
        averaged_state
    )

    return global_model
```

```python
# Create initial global model
global_model = create_model()

# Train one local model per hospital
client_models = []
client_sizes = []

for client_id, client_dataset in enumerate(
    client_datasets,
    start=1
):

    print(f"Training Hospital {client_id}...")

    local_model = train_local_model(
        global_model,
        client_dataset,
        local_epochs=3
    )

    client_models.append(local_model)
    client_sizes.append(len(client_dataset))

# Aggregate using FedAvg
global_model = fedavg(
    global_model,
    client_models,
    client_sizes
)

print("\nFedAvg round 1 completed successfully! 🚀")
```

```python
def train_fedprox_model(
    global_model,
    client_dataset,
    local_epochs=3,
    batch_size=32,
    learning_rate=0.001,
    mu=0.01
):
    # Copy global model to create the local model
    local_model = copy.deepcopy(global_model)

    local_loader = DataLoader(
        client_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    # Class weighting
    labels = client_dataset.tensors[1]

    num_positive = (labels == 1).sum().item()
    num_negative = (labels == 0).sum().item()

    client_pos_weight = torch.tensor(
        [num_negative / num_positive],
        dtype=torch.float32
    ).to(device)

    criterion = nn.BCEWithLogitsLoss(
        pos_weight=client_pos_weight
    )

    optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    # Save global parameters for proximal term
    global_params = {
        name: param.detach().clone()
        for name, param in global_model.named_parameters()
    }

    local_model.train()

    for epoch in range(local_epochs):

        for X_batch, y_batch in local_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            optimizer.zero_grad()

            # Local prediction
            outputs = local_model(X_batch)

            classification_loss = criterion(
                outputs,
                y_batch
            )

            # FedProx proximal term
            proximal_term = 0.0

            for name, param in local_model.named_parameters():

                proximal_term += torch.sum(
                    (param - global_params[name]) ** 2
                )

            proximal_term = (mu / 2) * proximal_term

            # Total FedProx loss
            loss = classification_loss + proximal_term

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                local_model.parameters(),
                max_norm=1.0
            )

            optimizer.step()
```

```python
import math
import copy


def adaptive_noise_multiplier(
    round_num,
    total_rounds,
    initial_noise=0.30,
    min_noise=0.05
):
    """
    Gradually decreases DP noise across federated rounds.
    """

    progress = (round_num - 1) / max(total_rounds - 1, 1)

    noise = initial_noise * math.exp(-2.0 * progress)

    return max(noise, min_noise)
```

```python
def clip_model_update(
    global_model,
    local_model,
    max_norm=1.0
):

    updates = {}

    total_norm_sq = 0.0

    # Calculate local - global update
    for (
        name,
        global_param
    ), (
        _,
        local_param
    ) in zip(
        global_model.named_parameters(),
        local_model.named_parameters()
    ):

        update = (
            local_param.detach()
            - global_param.detach()
        )

        updates[name] = update

        total_norm_sq += (
            update.norm(2).item() ** 2
        )

    total_norm = math.sqrt(
        total_norm_sq
    )

    # Clipping factor
    clip_factor = min(
        1.0,
        max_norm / (total_norm + 1e-12)
    )

    for name in updates:

        updates[name] = (
            updates[name] * clip_factor
        )

    return updates, total_norm, clip_factor
```

```python
def add_dp_noise(
    updates,
    noise_multiplier,
    max_norm=1.0
):

    noisy_updates = {}

    noise_std = (
        noise_multiplier * max_norm
    )

    for name, update in updates.items():

        noise = torch.randn_like(
            update
        ) * noise_std

        noisy_updates[name] = (
            update + noise
        )

    return noisy_updates
```

```python
def apply_noisy_updates(
    global_model,
    local_model,
    noise_multiplier,
    max_norm=1.0
):
    """
    Compute, clip, and noise a local model update.
    """

    updates, total_norm, clip_factor = clip_model_update(
        global_model,
        local_model,
        max_norm=max_norm
    )

    noisy_updates = add_dp_noise(
        updates,
        noise_multiplier=noise_multiplier,
        max_norm=max_norm
    )

    return noisy_updates, total_norm, clip_factor


def adaptive_dp_aggregate(
    global_model,
    local_models,
    client_sizes,
    noise_multiplier,
    max_norm=1.0
):
    """
    Aggregate clipped + noisy client updates.
    """

    total_samples = sum(client_sizes)

    global_state = global_model.state_dict()

    new_state = {}

    # Start from current global parameters
    for name, param in global_state.items():
        new_state[name] = param.detach().clone()

    # Aggregate noisy updates
    aggregated_updates = {
        name: torch.zeros_like(param)
        for name, param in global_state.items()
    }

    client_norms = []
    clip_factors = []

    for local_model, client_size in zip(
        local_models,
        client_sizes
    ):

        noisy_updates, total_norm, clip_factor = apply_noisy_updates(
            global_model,
            local_model,
            noise_multiplier=noise_multiplier,
            max_norm=max_norm
        )

        weight = client_size / total_samples

        for name in aggregated_updates:

            aggregated_updates[name] += (
                weight * noisy_updates[name]
            )

        client_norms.append(total_norm)
        clip_factors.append(clip_factor)

    # Apply aggregated update
    for name in new_state:

        new_state[name] = (
            new_state[name]
            + aggregated_updates[name]
        )

    global_model.load_state_dict(new_state)

    return (
        global_model,
        client_norms,
        clip_factors
    )
```

```python
class DPFTTransformer(nn.Module):
    def __init__(
        self,
        num_features=11,
        d_token=64,
        n_heads=4,
        n_layers=3,
        dropout=0.2
    ):
        super().__init__()

        self.num_features = num_features
        self.d_token = d_token

        # Feature embeddings
        self.feature_embeddings = nn.Parameter(
            torch.randn(num_features, d_token) * 0.02
        )

        # CLS token
        self.cls_token = nn.Parameter(
            torch.randn(1, 1, d_token) * 0.02
        )

        # DP-compatible attention layers
        self.attention_layers = nn.ModuleList()

        for _ in range(n_layers):

            self.attention_layers.append(
                nn.ModuleDict({
                    "norm1": nn.LayerNorm(d_token),

                    "attention": DPMultiheadAttention(
                        embed_dim=d_token,
                        num_heads=n_heads,
                        dropout=dropout,
                        batch_first=True
                    ),

                    "norm2": nn.LayerNorm(d_token),

                    "ffn": nn.Sequential(
                        nn.Linear(
                            d_token,
                            d_token * 4
                        ),
                        nn.GELU(),
                        nn.Dropout(dropout),
                        nn.Linear(
                            d_token * 4,
                            d_token
                        ),
                        nn.Dropout(dropout)
                    )
                })
            )

        # Final normalization
        self.norm = nn.LayerNorm(d_token)

        # Classification head
        self.classifier = nn.Sequential(
            nn.Linear(d_token, 32),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(32, 1)
        )

    def forward(self, x):

        # Numerical features → feature tokens
        x = (
            x.unsqueeze(-1)
            * self.feature_embeddings.unsqueeze(0)
        )

        batch_size = x.size(0)

        # CLS token
        cls = self.cls_token.expand(
            batch_size,
            -1,
            -1
        )

        x = torch.cat(
            [cls, x],
            dim=1
        )

        # Transformer layers
        for layer in self.attention_layers:

            # Pre-normalization
            residual = x

            x_norm = layer["norm1"](x)

            attention_output, _ = layer["attention"](
                x_norm,
                x_norm,
                x_norm,
                need_weights=False
            )

            x = residual + attention_output

            # Feed-forward network
            residual = x

            x_norm = layer["norm2"](x)

            x = residual + layer["ffn"](x_norm)

        # CLS representation
        x = self.norm(x[:, 0])

        # Binary classification
        output = self.classifier(x)

        return output.squeeze(1)


print("DP-compatible FT-Transformer created successfully!")
```

```python
from opacus import PrivacyEngine


def train_formal_dp_local(
    global_model,
    client_dataset,
    noise_multiplier=1.0,
    max_grad_norm=1.0,
    batch_size=32,
    local_epochs=1,
    learning_rate=0.001
):
    # Create a fresh copy of the global model
    local_model = copy.deepcopy(global_model)

    # Local DataLoader
    local_loader = DataLoader(
        client_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    # Class weighting
    labels = client_dataset.tensors[1]

    num_positive = (labels == 1).sum().item()
    num_negative = (labels == 0).sum().item()

    pos_weight = torch.tensor(
        [num_negative / num_positive],
        dtype=torch.float32
    ).to(device)

    criterion = nn.BCEWithLogitsLoss(
        pos_weight=pos_weight
    )

    optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    # Create privacy engine
    privacy_engine = PrivacyEngine()

    local_model, optimizer, local_loader = (
        privacy_engine.make_private(
            module=local_model,
            optimizer=optimizer,
            data_loader=local_loader,
            noise_multiplier=noise_multiplier,
            max_grad_norm=max_grad_norm
        )
    )

    # Local DP-SGD training
    local_model.train()

    for epoch in range(local_epochs):

        for X_batch, y_batch in local_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            optimizer.zero_grad()

            outputs = local_model(X_batch)

            loss = criterion(
                outputs,
                y_batch
            )

            loss.backward()

            optimizer.step()

    return (
        local_model,
        privacy_engine,
        len(local_loader)
    )
```

```python
def dp_fedavg(
    global_model,
    client_models,
    client_sizes
):
    """
    FedAvg aggregation for Opacus-trained models.

    Uses the underlying module of each GradSampleModule.
    """

    total_samples = sum(client_sizes)

    global_state = global_model.state_dict()

    averaged_state = {
        key: torch.zeros_like(value)
        for key, value in global_state.items()
    }

    for client_model, client_size in zip(
        client_models,
        client_sizes
    ):

        # Get underlying model if wrapped by Opacus
        if hasattr(client_model, "_module"):
            client_state = (
                client_model._module.state_dict()
            )
        else:
            client_state = client_model.state_dict()

        weight = (
            client_size / total_samples
        )

        for key in averaged_state:

            averaged_state[key] += (
                client_state[key].detach()
                * weight
            )

    global_model.load_state_dict(
        averaged_state
    )

    return global_model
```
